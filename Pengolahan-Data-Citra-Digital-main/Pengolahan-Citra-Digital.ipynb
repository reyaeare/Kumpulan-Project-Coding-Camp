## 1. Pendahuluan

**Pengolahan Citra Digital (Digital Image Processing)** adalah bidang ilmu yang mempelajari bagaimana suatu citra (gambar) dibentuk, diolah, dan dianalisis menggunakan komputer sehingga menghasilkan informasi yang dapat dipahami oleh manusia atau dapat digunakan oleh sistem lain (misalnya model machine learning).

Secara sederhana, sebuah **citra digital** adalah representasi gambar dalam bentuk matriks angka. Setiap elemen matriks disebut **piksel (pixel)**, yang menyimpan nilai intensitas warna.

- **Citra Grayscale**: setiap piksel hanya punya 1 nilai (0–255), merepresentasikan derajat keabuan (0 = hitam, 255 = putih).
- **Citra RGB (berwarna)**: setiap piksel punya 3 nilai (Red, Green, Blue), masing-masing 0–255.
- **Citra Biner**: setiap piksel hanya bernilai 0 (hitam) atau 255 (putih).

## 2. Dasar Pemrograman Python untuk Pengolahan Citra

Sebelum masuk ke pengolahan citra, kita perlu memahami bahwa **citra = array/matriks angka**. Bagian ini mengulas cara Python merepresentasikan data numerik yang nanti dipakai untuk merepresentasikan piksel gambar.

### Array 1D, 2D, dan 3D
- Array 1D → seperti daftar angka (list)
- Array 2D → seperti tabel/matriks (dipakai untuk citra grayscale: `[tinggi, lebar]`)
- Array 3D → matriks dengan channel warna (dipakai untuk citra RGB: `[tinggi, lebar, channel]`)


```plaintext
# Contoh array 1D
array_1d = np.array([10, 20, 30, 40, 50])
print("Array 1D:", array_1d)

# Contoh array 2D -> analog dengan citra grayscale kecil (3x3 piksel)
array_2d = np.array([
    [0, 128, 255],
    [64, 192, 32],
    [10, 220, 100]
])
print("\nArray 2D (analog citra grayscale 3x3):\n", array_2d)
print("Ukuran (tinggi, lebar):", array_2d.shape)

# Contoh array 3D -> analog dengan citra RGB (2x2 piksel, 3 channel warna)
array_3d = np.array([
    [[255, 0, 0], [0, 255, 0]],
    [[0, 0, 255], [255, 255, 0]]
])
print("\nArray 3D (analog citra RGB 2x2):\n", array_3d)
print("Ukuran (tinggi, lebar, channel):", array_3d.shape)
```

⚠️ **Penting:** OpenCV membaca gambar dengan urutan channel **BGR**, bukan RGB seperti Pillow/Matplotlib. Jadi setiap kali membaca gambar dengan `cv2.imread()`, perlu dikonversi dulu ke RGB sebelum ditampilkan dengan matplotlib.


```plaintext
# Ganti path berikut dengan gambar milik kamu sendiri
path_gambar = 'gambar.jpg'

# ---- Membaca dengan Pillow ----
try:
    citra_pil = Image.open(path_gambar)
    print("Berhasil dibuka dengan Pillow")
    print("Ukuran (lebar, tinggi):", citra_pil.size)
    print("Mode warna:", citra_pil.mode)
except FileNotFoundError:
    print("File belum ada. Silakan upload gambar terlebih dahulu lalu ganti 'path_gambar'.")
```

```plaintext
# ---- Konversi ke Grayscale ----
if citra_cv is not None:
    citra_gray = cv2.cvtColor(citra_cv, cv2.COLOR_BGR2GRAY)
    plt.imshow(citra_gray, cmap='gray')
    plt.title("Citra Grayscale")
    plt.axis('off')
    plt.show()
    print("Ukuran citra grayscale:", citra_gray.shape)  # hanya (tinggi, lebar), tanpa channel
```

## 5. Histogram Citra

**Histogram citra** adalah grafik yang menunjukkan distribusi/sebaran nilai intensitas piksel dalam sebuah citra. Sumbu X = nilai intensitas (0–255), sumbu Y = jumlah piksel yang memiliki nilai tersebut.

Histogram berguna untuk:
- Menganalisis kecerahan & kontras citra
- Menentukan nilai ambang (threshold) untuk binerisasi
- Deteksi citra under-exposed / over-exposed

Histogram condong ke kiri → citra gelap. Histogram condong ke kanan → citra terang.


```plaintext
if citra_cv is not None:
    plt.figure(figsize=(8, 4))
    plt.hist(citra_gray.ravel(), bins=256, range=(0, 255), color='gray')
    plt.title("Histogram Citra Grayscale")
    plt.xlabel("Nilai Intensitas Piksel")
    plt.ylabel("Jumlah Piksel")
    plt.show()
```

```plaintext
# Histogram per channel warna (R, G, B)
if citra_cv is not None:
    warna = ('r', 'g', 'b')
    plt.figure(figsize=(8, 4))
    for i, c in enumerate(warna):
        hist = cv2.calcHist([citra_cv_rgb], [i], None, [256], [0, 256])
        plt.plot(hist, color=c, label=f'Channel {c.upper()}')
    plt.title("Histogram per Channel Warna (RGB)")
    plt.xlabel("Nilai Intensitas")
    plt.ylabel("Jumlah Piksel")
    plt.legend()
    plt.show()
```

## 6. Pengolahan Dasar Citra

Bagian ini membahas operasi-operasi citra paling fundamental:

1. **Citra Negatif** → membalik nilai intensitas piksel (`255 - nilai_piksel`)
2. **Pencerahan/Kecerahan (Brightness)** → menambah/mengurangi nilai piksel
3. **Citra Biner (Binerisasi/Thresholding)** → mengubah citra grayscale menjadi hitam-putih murni berdasarkan nilai ambang


```plaintext
# ---- 1. Citra Negatif ----
if citra_cv is not None:
    citra_negatif = 255 - citra_cv_rgb

    fig, axes = plt.subplots(1, 2, figsize=(10, 5))
    axes[0].imshow(citra_cv_rgb); axes[0].set_title("Citra Asli"); axes[0].axis('off')
    axes[1].imshow(citra_negatif); axes[1].set_title("Citra Negatif"); axes[1].axis('off')
    plt.show()
```

```plaintext
# ---- 2. Fungsi Pencerahan Citra (Brightness) ----
def atur_kecerahan(citra, nilai):
    '''
    Menambah/mengurangi kecerahan citra.
    nilai positif -> lebih terang, nilai negatif -> lebih gelap
    np.clip digunakan agar nilai piksel tetap berada di rentang 0-255
    '''
    citra_baru = citra.astype(np.int16) + nilai
    citra_baru = np.clip(citra_baru, 0, 255).astype(np.uint8)
    return citra_baru

if citra_cv is not None:
    citra_terang = atur_kecerahan(citra_cv_rgb, 60)
    citra_gelap = atur_kecerahan(citra_cv_rgb, -60)

    fig, axes = plt.subplots(1, 3, figsize=(12, 4))
    axes[0].imshow(citra_gelap); axes[0].set_title("Lebih Gelap (-60)"); axes[0].axis('off')
    axes[1].imshow(citra_cv_rgb); axes[1].set_title("Asli"); axes[1].axis('off')
    axes[2].imshow(citra_terang); axes[2].set_title("Lebih Terang (+60)"); axes[2].axis('off')
    plt.show()
```

```plaintext
# ---- 3. Citra Biner (Thresholding) ----
if citra_cv is not None:
    ambang = 127  # nilai ambang/threshold, bisa disesuaikan (0-255)
    _, citra_biner = cv2.threshold(citra_gray, ambang, 255, cv2.THRESH_BINARY)

    fig, axes = plt.subplots(1, 2, figsize=(10, 5))
    axes[0].imshow(citra_gray, cmap='gray'); axes[0].set_title("Grayscale"); axes[0].axis('off')
    axes[1].imshow(citra_biner, cmap='gray'); axes[1].set_title(f"Citra Biner (ambang={ambang})"); axes[1].axis('off')
    plt.show()
```

## 7. Filter & Konvolusi

**Konvolusi** adalah operasi matematis di mana sebuah **kernel/mask/filter** (matriks kecil, misal 3x3) "digeser" ke seluruh bagian citra untuk menghasilkan efek tertentu.

Beberapa filter umum:
| Filter | Fungsi |
|---|---|
| Mean/Averaging Blur | Menghaluskan citra (mengurangi noise) |
| Gaussian Blur | Blur yang lebih natural, bobot kernel mengikuti distribusi normal |
| Median Blur | Blur yang efektif menghilangkan noise "salt & pepper" |
| Sharpening | Mempertajam detail/tepi citra |
| Sobel/Canny | Deteksi tepi (edge detection) |


```plaintext
if citra_cv is not None:
    # Mean Blur (rata-rata sederhana)
    blur_mean = cv2.blur(citra_cv_rgb, (5, 5))

    # Gaussian Blur
    blur_gaussian = cv2.GaussianBlur(citra_cv_rgb, (5, 5), 0)

    # Median Blur
    blur_median = cv2.medianBlur(citra_cv_rgb, 5)

    fig, axes = plt.subplots(1, 4, figsize=(16, 4))
    axes[0].imshow(citra_cv_rgb); axes[0].set_title("Asli"); axes[0].axis('off')
    axes[1].imshow(blur_mean); axes[1].set_title("Mean Blur"); axes[1].axis('off')
    axes[2].imshow(blur_gaussian); axes[2].set_title("Gaussian Blur"); axes[2].axis('off')
    axes[3].imshow(blur_median); axes[3].set_title("Median Blur"); axes[3].axis('off')
    plt.show()
```

```plaintext
# ---- Sharpening (mempertajam citra) ----
if citra_cv is not None:
    kernel_sharpen = np.array([
        [ 0, -1,  0],
        [-1,  5, -1],
        [ 0, -1,  0]
    ])
    citra_tajam = cv2.filter2D(citra_cv_rgb, -1, kernel_sharpen)

    fig, axes = plt.subplots(1, 2, figsize=(10, 5))
    axes[0].imshow(citra_cv_rgb); axes[0].set_title("Asli"); axes[0].axis('off')
    axes[1].imshow(citra_tajam); axes[1].set_title("Citra Dipertajam (Sharpen)"); axes[1].axis('off')
    plt.show()
```

```plaintext
# ---- Deteksi Tepi (Edge Detection) ----
if citra_cv is not None:
    tepi_sobel_x = cv2.Sobel(citra_gray, cv2.CV_64F, 1, 0, ksize=3)
    tepi_sobel_y = cv2.Sobel(citra_gray, cv2.CV_64F, 0, 1, ksize=3)
    tepi_canny = cv2.Canny(citra_gray, 100, 200)

    fig, axes = plt.subplots(1, 3, figsize=(14, 5))
    axes[0].imshow(np.abs(tepi_sobel_x), cmap='gray'); axes[0].set_title("Sobel X"); axes[0].axis('off')
    axes[1].imshow(np.abs(tepi_sobel_y), cmap='gray'); axes[1].set_title("Sobel Y"); axes[1].axis('off')
    axes[2].imshow(tepi_canny, cmap='gray'); axes[2].set_title("Canny Edge Detection"); axes[2].axis('off')
    plt.show()
```

## 8. Contrast Enhancement & Gray Level Slicing

### Contrast Stretching
Meregangkan rentang nilai intensitas piksel agar mencakup rentang penuh (0–255), sehingga kontras citra meningkat.

### Histogram Equalization
Teknik meratakan distribusi histogram sehingga detail citra (terutama yang gelap/terlalu terang) menjadi lebih terlihat.

### Gray Level Slicing
Menonjolkan (highlight) rentang nilai intensitas tertentu, sementara nilai lain "diredam".


```plaintext
if citra_cv is not None:
    # ---- Histogram Equalization ----
    citra_equalized = cv2.equalizeHist(citra_gray)

    fig, axes = plt.subplots(1, 2, figsize=(10, 5))
    axes[0].imshow(citra_gray, cmap='gray'); axes[0].set_title("Grayscale Asli"); axes[0].axis('off')
    axes[1].imshow(citra_equalized, cmap='gray'); axes[1].set_title("Setelah Histogram Equalization"); axes[1].axis('off')
    plt.show()

    fig, axes = plt.subplots(1, 2, figsize=(10, 3))
    axes[0].hist(citra_gray.ravel(), bins=256, range=(0,255), color='gray'); axes[0].set_title("Histogram Sebelum")
    axes[1].hist(citra_equalized.ravel(), bins=256, range=(0,255), color='gray'); axes[1].set_title("Histogram Sesudah")
    plt.show()
```

```plaintext
# ---- Gray Level Slicing ----
def gray_level_slicing(citra, batas_bawah, batas_atas, nilai_highlight=255):
    '''
    Menonjolkan piksel yang nilainya berada di antara batas_bawah dan batas_atas.
    Piksel lain tetap menggunakan nilai aslinya (bisa juga diset ke 0 tergantung kebutuhan).
    '''
    hasil = citra.copy()
    mask = (citra >= batas_bawah) & (citra <= batas_atas)
    hasil[mask] = nilai_highlight
    return hasil

if citra_cv is not None:
    citra_sliced = gray_level_slicing(citra_gray, 100, 180)

    fig, axes = plt.subplots(1, 2, figsize=(10, 5))
    axes[0].imshow(citra_gray, cmap='gray'); axes[0].set_title("Grayscale Asli"); axes[0].axis('off')
    axes[1].imshow(citra_sliced, cmap='gray'); axes[1].set_title("Gray Level Slicing (100-180)"); axes[1].axis('off')
    plt.show()
```

## 9. Operasi Morfologi Citra

Operasi morfologi umumnya diterapkan pada **citra biner** untuk memperbaiki bentuk objek hasil segmentasi. Operasi dasar:

- **Erosi (Erosion)** → "mengikis" tepi objek putih, mengecilkan objek, menghilangkan noise kecil
- **Dilasi (Dilation)** → "menebalkan" objek putih, menyambung bagian yang terputus
- **Opening** (Erosi lalu Dilasi) → menghilangkan noise kecil tanpa mengubah ukuran objek utama
- **Closing** (Dilasi lalu Erosi) → menutup lubang kecil di dalam objek


```plaintext
if citra_cv is not None:
    kernel = np.ones((5, 5), np.uint8)

    citra_erosi = cv2.erode(citra_biner, kernel, iterations=1)
    citra_dilasi = cv2.dilate(citra_biner, kernel, iterations=1)
    citra_opening = cv2.morphologyEx(citra_biner, cv2.MORPH_OPEN, kernel)
    citra_closing = cv2.morphologyEx(citra_biner, cv2.MORPH_CLOSE, kernel)

    fig, axes = plt.subplots(1, 5, figsize=(18, 4))
    judul = ["Biner Asli", "Erosi", "Dilasi", "Opening", "Closing"]
    hasil = [citra_biner, citra_erosi, citra_dilasi, citra_opening, citra_closing]
    for ax, img, t in zip(axes, hasil, judul):
        ax.imshow(img, cmap='gray'); ax.set_title(t); ax.axis('off')
    plt.show()
```

## 10. Segmentasi & Thresholding Lanjutan

**Segmentasi citra** bertujuan memisahkan objek dari background. Selain thresholding sederhana (binary threshold), ada teknik yang lebih adaptif:

- **Otsu's Thresholding** → menentukan nilai ambang optimal secara otomatis berdasarkan histogram citra
- **Adaptive Thresholding** → nilai ambang dihitung per-wilayah kecil, cocok untuk citra dengan pencahayaan tidak merata


```plaintext
if citra_cv is not None:
    # Otsu's Thresholding (ambang otomatis)
    ambang_otsu, citra_otsu = cv2.threshold(citra_gray, 0, 255, cv2.THRESH_BINARY + cv2.THRESH_OTSU)
    print(f"Nilai ambang optimal (Otsu): {ambang_otsu}")

    # Adaptive Thresholding
    citra_adaptive = cv2.adaptiveThreshold(
        citra_gray, 255, cv2.ADAPTIVE_THRESH_GAUSSIAN_C, cv2.THRESH_BINARY, 11, 2
    )

    fig, axes = plt.subplots(1, 3, figsize=(14, 5))
    axes[0].imshow(citra_gray, cmap='gray'); axes[0].set_title("Grayscale"); axes[0].axis('off')
    axes[1].imshow(citra_otsu, cmap='gray'); axes[1].set_title(f"Otsu (ambang={int(ambang_otsu)})"); axes[1].axis('off')
    axes[2].imshow(citra_adaptive, cmap='gray'); axes[2].set_title("Adaptive Threshold"); axes[2].axis('off')
    plt.show()
```

## 11. Studi Kasus — Preprocessing Citra untuk Klasifikasi (Contoh: Klasifikasi Sampah)

Pada kasus nyata seperti **klasifikasi citra sampah** (Recyclable, Electronic, Organic) untuk keperluan machine learning/deep learning, tahapan preprocessing yang umum dilakukan meliputi:

1. **Resize** → menyamakan ukuran seluruh citra (misal 224x224 untuk arsitektur seperti MobileNetV2/EfficientNet)
2. **Normalisasi** → mengubah rentang nilai piksel dari 0–255 menjadi 0–1 (atau sesuai kebutuhan arsitektur)
3. **Augmentasi Data** → memperbanyak variasi data secara sintetis (flip, rotate, brightness) agar model lebih general dan tidak overfitting
4. **Denoising (opsional)** → menghilangkan noise agar fitur citra lebih bersih

Kode di bawah menunjukkan pipeline sederhana preprocessing yang bisa diadaptasi untuk dataset klasifikasi citra.


```plaintext
def preprocessing_untuk_klasifikasi(path_gambar, ukuran=(224, 224)):
    '''
    Pipeline dasar preprocessing citra sebelum masuk ke model klasifikasi.
    1. Baca citra
    2. Konversi BGR -> RGB
    3. Resize ke ukuran target
    4. Normalisasi ke rentang 0-1
    '''
    citra = cv2.imread(path_gambar)
    if citra is None:
        raise FileNotFoundError(f"Gambar tidak ditemukan di path: {path_gambar}")

    citra = cv2.cvtColor(citra, cv2.COLOR_BGR2RGB)
    citra_resize = cv2.resize(citra, ukuran)
    citra_normalisasi = citra_resize.astype(np.float32) / 255.0

    return citra_resize, citra_normalisasi

if citra_cv is not None:
    citra_resize, citra_norm = preprocessing_untuk_klasifikasi(path_gambar)
    print("Ukuran setelah resize:", citra_resize.shape)
    print("Rentang nilai piksel setelah normalisasi: min =", citra_norm.min(), ", max =", citra_norm.max())

    plt.imshow(citra_resize)
    plt.title("Citra Setelah Resize (siap untuk augmentasi)")
    plt.axis('off')
    plt.show()
```

```plaintext
# ---- Contoh Augmentasi Sederhana secara Manual ----
def augmentasi_flip_horizontal(citra):
    return cv2.flip(citra, 1)

def augmentasi_rotasi(citra, sudut):
    (h, w) = citra.shape[:2]
    pusat = (w // 2, h // 2)
    matriks_rotasi = cv2.getRotationMatrix2D(pusat, sudut, 1.0)
    return cv2.warpAffine(citra, matriks_rotasi, (w, h))

if citra_cv is not None:
    citra_flip = augmentasi_flip_horizontal(citra_resize)
    citra_rotasi = augmentasi_rotasi(citra_resize, 25)

    fig, axes = plt.subplots(1, 3, figsize=(12, 4))
    axes[0].imshow(citra_resize); axes[0].set_title("Asli"); axes[0].axis('off')
    axes[1].imshow(citra_flip); axes[1].set_title("Flip Horizontal"); axes[1].axis('off')
    axes[2].imshow(citra_rotasi); axes[2].set_title("Rotasi 25°"); axes[2].axis('off')
    plt.show()

print("\n💡 Untuk augmentasi skala besar pada proyek nyata (misalnya kompetisi klasifikasi citra),")
print("lebih disarankan menggunakan ImageDataGenerator (Keras) atau library albumentations")
print("agar augmentasi dilakukan otomatis & efisien saat training model.")
```
